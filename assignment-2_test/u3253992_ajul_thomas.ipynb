# Overall Results Plot


````python
# plots overall results of the genetic algorithm

import matplotlib.pyplot as plt

# Data for the scatter plot
average_wait_time = [
    40.45328,
    38.4595,
    38.6992,
    43.752,
    42.3325,
    38.4211,
    63.3025,
    69.14,
    75.14,
]
max_wait_time = [631, 746, 722, 526, 777, 636, 808, 826, 715]

# Find the minimum values for average wait time and max wait time
min_avg_wait_time = min(average_wait_time)
min_max_wait_time = min(max_wait_time)

# Create scatter plot with horizontal and vertical lines at the lowest points
plt.figure(figsize=(8, 6))
plt.scatter(average_wait_time, max_wait_time, color="blue", marker="o")
plt.axhline(
    y=min_max_wait_time,
    color="red",
    linestyle="--",
    label=f"Min Max Wait Time = {min_max_wait_time}",
)
plt.axvline(
    x=min_avg_wait_time,
    color="green",
    linestyle="--",
    label=f"Min Avg Wait Time = {min_avg_wait_time}",
)
plt.title("Scatter Plot of Average Wait Time vs Max Wait Time with Minimum Lines")
plt.xlabel("Average Wait Time")
plt.ylabel("Max Wait Time")
plt.legend()
plt.grid(True)
plt.show()
````
